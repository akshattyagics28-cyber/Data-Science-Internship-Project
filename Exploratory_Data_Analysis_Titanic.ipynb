```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='deep')
plt.rcParams['figure.figsize'] = (9, 5)
```

## 2. Load the Dataset
No dataset was provided with the assignment, so this project uses a Titanic-style educational sample. The file contains realistic passenger fields and a few missing values and duplicate records for investigation.


```python
df = pd.read_csv('dataset/titanic_eda_sample.csv')
df.head()
```

## 3. Dataset Overview


```python
print('Rows and columns:', df.shape)
print('\nColumn names:')
print(df.columns.tolist())
print('\nData types:')
print(df.dtypes)
print('\nFirst five rows:')
df.head()
```

## 4. Statistical Summary
This provides the count, mean, standard deviation, minimum, maximum, and quartile values for numerical columns.


```python
df.describe()
```

## 5. Data Quality Check


```python
print('Missing values by column:')
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())

plt.figure(figsize=(8, 4))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False, cmap='viridis')
plt.title('Missing Values Map')
plt.show()
```

## 6. Univariate Analysis
Univariate analysis examines one variable at a time. Here we inspect survival, passenger class, age, and fare.


```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
sns.countplot(data=df, x='Survived', ax=axes[0])
axes[0].set_xticklabels(['Did Not Survive', 'Survived'])
axes[0].set_title('Passenger Survival Count')
sns.countplot(data=df, x='Pclass', ax=axes[1])
axes[1].set_title('Passenger Count by Class')
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))
sns.histplot(data=df, x='Age', bins=20, kde=True, color='#7c3aed')
plt.title('Age Distribution')
plt.show()
```

## 7. Bivariate Analysis
Bivariate analysis compares two variables to identify relationships.


```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
sns.countplot(data=df, x='Sex', hue='Survived', ax=axes[0])
axes[0].set_title('Survival by Gender')
sns.countplot(data=df, x='Pclass', hue='Survived', ax=axes[1])
axes[1].set_title('Survival by Passenger Class')
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Pclass', y='Fare', hue='Pclass', legend=False, palette='Set2')
plt.title('Fare Distribution by Passenger Class')
plt.show()
```

## 8. Correlation Analysis
Correlation measures the direction and strength of a linear relationship between numerical variables. Values close to +1 or -1 show stronger relationships.


```python
numeric_columns = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
correlation = df[numeric_columns].corr()
correlation

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

## 9. Key Findings

1. **Gender is an important factor.** Female passengers show a higher survival rate than male passengers in this dataset.
2. **Passenger class affects outcomes.** First-class passengers generally have a higher survival rate than third-class passengers.
3. **Fare is linked with class.** Higher fares are more common among first-class passengers.
4. **Age has a wide distribution.** Most passengers are young adults, while there are fewer elderly passengers.
5. **Data quality needs attention.** The raw data contains missing values and duplicate records, which should be handled before predictive modeling.

## 10. Conclusion
This EDA project used statistical summaries and visualizations to explore passenger survival. The analysis shows that gender, passenger class, and fare are key influencing factors. The findings can guide further data cleaning and machine-learning work.
